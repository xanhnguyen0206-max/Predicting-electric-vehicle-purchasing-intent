#### 📌 Mất cân bằng Nhãn Mục tiêu (Class Imbalance)
* **Nó là gì?**  
  Hiện tượng số lượng quan sát của các lớp trong bài toán phân loại chênh lệch nhau rất lớn. Trong bài toán này:
  - Lớp đa số (Negative Class - `No: 0`): **551.886 mẫu (82,54%)**.
  - Lớp thiểu số (Positive Class - `Yes: 1`): **116.779 mẫu (17,46%)**.
  - Tỷ lệ mất cân bằng: $551.886 / 116.779 \approx 4,7259 : 1 \approx 4,73 : 1$.
* **Cạm bẫy chết người (The Accuracy Paradox):**  
  Nếu một mô hình "ngớ ngẩn" dự đoán 100% khách hàng đều là `No` (Không mua), mô hình đó vẫn đạt **Accuracy = 82,54%**! Tuy nhiên, mô hình hoàn toàn vô dụng vì bỏ sót 100% khách hàng thực sự mua xe (Recall = 0%).  
  $\Rightarrow$ Đây là lý do cốt lõi buộc chúng tôi phải sử dụng trọng số phạt `class_weight='balanced'` và `scale_pos_weight=4.73`, đồng thời lấy **PR-AUC** làm kim chỉ nam thay vì Accuracy!

### 2. Từ điển Cú pháp SQL & Khái niệm Chuyên ngành trong Cell 7 & 8

#### 📌 Biểu thức Điều kiện `CASE WHEN ... THEN ... ELSE ... END`
* **Nó là gì?**  
  Cú pháp rẽ nhánh điều kiện trong SQL tương tự như `if-else` trong lập trình.
* **Tại sao dùng?**  
  Dùng để rời rạc hóa biến số liên tục (Binning) thành các nhóm thu nhập hoặc nhóm cự ly có ý nghĩa nghiệp vụ (ví dụ: chia thu nhập thành các khoảng Dưới 50k, 50k - 80k, 80k - 110k, Trên 110k).

#### 📌 Kỹ thuật Tính Tỷ lệ Chuyển đổi bằng `AVG(CASE WHEN ... THEN 1.0 ELSE 0.0 END)`
* **Nó là gì?**  
  Một thủ thuật SQL kinh điển để tính xác suất hoặc tỷ lệ phần trăm của nhãn phân loại:
  ```sql
  AVG(CASE WHEN Will_Buy_EV = 'Yes' THEN 1.0 ELSE 0.0 END) * 100 AS Ty_Le_Mua_EV_Pct
  ```
* **Bản chất toán học:**  
  Nếu khách hàng mua xe (`Yes`), gán giá trị 1.0; nếu không mua (`No`), gán 0.0. Trung bình cộng của dãy số gồm các số 0 và 1 chính là **Tỷ lệ phần trăm xuất hiện của số 1** (tương đương $\frac{\sum \text{Yes}}{N}$).

---

### 2. Từ điển Thuật ngữ Toán học & Thống kê trong Cell 10 & 11

#### 📌 Hệ số Tương quan Tuyến tính Pearson ($r$)
* **Nó là gì?**  
  Thước đo đo lường mức độ mạnh yếu của mối quan hệ tuyến tính giữa hai biến số liên tục $X$ và $Y$:
  $$r = \frac\frac{\sum_{i=1}^n (x_i - \bar{x})(y_i - \bar{y})}{\sqrt{\sum_{i=1}^n (x_i - \bar{x})^2} \sqrt{\sum_{i=1}^n (y_i - \bar{y})^2}} \in [-1, 1]$$
  - $r = +1$: Tương quan dương hoàn hảo (biến này tăng thì biến kia tăng tuyến tính tuyệt đối).
  - $r = -1$: Tương quan âm hoàn hảo.
  - $r = 0$: Không có mối liên hệ tuyến tính.
* **Hiện tượng trong dự án xe điện:**  
  Tất cả các hệ số tương quan tuyến tính giữa các biến đầu vào với biến mục tiêu `Will_Buy_EV` đều có giá trị tuyệt đối rất nhỏ: $|r| < 0,25$.
* **Ý nghĩa quyết định phương pháp:**  
  Nhiều người lầm tưởng $|r|$ thấp nghĩa là các biến không quan trọng. Thực chất, điều này khẳng định **mối quan hệ giữa hành vi mua xe điện và các yếu tố đời sống là mối quan hệ phi tuyến tính phức tạp (Non-linear Interactions)**. Nếu chỉ sử dụng các mô hình tuyến tính đơn giản, hiệu năng sẽ bị giới hạn; bắt buộc phải sử dụng các thuật toán dạng cây (Cây quyết định, Random Forest, LightGBM, XGBoost) có khả năng phân nhánh phi tuyến!

#### 📌 Hiện tượng Đa cộng tuyến (Multicollinearity)
* **Nó là gì?**  
  Tình trạng hai hoặc nhiều biến độc lập trong mô hình hồi quy có mối quan hệ tương quan tuyến tính rất chặt chẽ với nhau (ví dụ: Biến $X_2 pprox 2 \cdot X_1 + 5$).
* **Hậu quả nguy hại:**  
  - Làm cho ma trận nghịch đảo $(X^T X)^{-1}$ tiến sát điểm suy biến.
  - Phương sai của các hệ số ước lượng bị thổi phồng cực lớn, khiến dấu của hệ số bị đảo lộn vô lý và mô hình trở nên cực kỳ bất ổn trước những thay đổi nhỏ của dữ liệu.

#### 📌 Hệ số Phóng đại Phương sai (Variance Inflation Factor - VIF)
* **Nó là gì?**  
  Chỉ số đo lường mức độ mà phương sai của một hệ số hồi quy bị thổi phồng do mối quan hệ phụ thuộc tuyến tính vào các biến độc lập khác:
  $$\text{VIF}_j = \frac\frac{1}{1 - R_j^2}$$
  Trong đó $R_j^2$ là hệ số xác định (R-squared) thu được khi thực hiện hồi quy biến độc lập $X_j$ theo toàn bộ các biến độc lập còn lại.
* **Quy chuẩn đánh giá VIF:**
  - $\text{VIF} = 1$: Hoàn toàn độc lập, không có đa cộng tuyến.
  - $1 < \text{VIF} < 5$: Mức độ tương quan chấp nhận được, an toàn tuyệt đối.
  - $5 \le \text{VIF} < 10$: Có dấu hiệu đa cộng tuyến cần theo dõi.
  - $\text{VIF} \ge 10$: Đa cộng tuyến nghiêm trọng, bắt buộc phải loại bỏ hoặc gộp biến.

### 2. Bóc tách 5 Đặc trưng Tương tác Phi tuyến Sáng tạo trong Cell 14

#### ⚙️ Đặc trưng 1: Gánh nặng Chi phí Đi lại trên Thu nhập (`Commute_Burden`)
$$\text{Commute\_Burden} = \frac\frac{\text{Daily\_Commute\_km}}{\text{Annual\_Income\_USD} + 1000.0}$$
* **Ý nghĩa:** Một người thu nhập thấp nhưng phải lái xe 60 km/ngày chịu gánh nặng chi phí xăng dầu rất lớn, do đó động lực tiết kiệm chi phí vận hành bằng xe điện sẽ cao hơn người thu nhập cao nhưng chỉ đi 5 km/ngày.
* **Tại sao cộng thêm 1000.0?** Đây là kỹ thuật làm mịn (Smoothing constant) giúp tránh lỗi chia cho 0 và ngăn chặn việc thu nhập quá nhỏ làm phát sinh giá trị vô cực.

#### ⚙️ Đặc trưng 2: Điểm Tổng hợp Hạ tầng Sạc (`Charging_Infrastructure_Score`)
$$\text{Charging\_Infrastructure\_Score} = (\text{Charging\_Stations\_Near\_Home} \times \mathbb{I}_{\text{Home\_Charging}}) + \text{Charging\_Stations\_Near\_Work}$$
* **Ý nghĩa:** Nếu có khả năng sạc tại nhà ($\mathbb{I}_{\text{Home\_Charging}} = 1$), số trạm sạc quanh nhà sẽ phát huy giá trị tối đa; cộng thêm số trạm sạc quanh cơ quan giúp mô hình định lượng toàn diện mức độ tiện lợi tiếp cận nguồn điện của khách hàng.

#### ⚙️ Đặc trưng 3: Tương tác Đòn bẩy Trợ cấp - Thu nhập (`Subsidy_Income_Interaction`)
$$\text{Subsidy\_Income\_Interaction} = \mathbb{I}_{\text{Subsidy\_Available}} \times \text{Annual\_Income\_USD}$$
* **Ý nghĩa:** Chính sách trợ giá có tác động kích thích tâm lý khác nhau tùy theo mức thu nhập của khách hàng.

#### ⚙️ Đặc trưng 4: Tương tác Ý thức Môi trường - Cự ly Di chuyển (`Env_Commute_Interaction`)
$$\text{Env\_Commute\_Interaction} = \text{Environmental\_Concern\_Level} \times \text{Daily\_Commute\_km}$$
* **Ý nghĩa:** Người có ý thức môi trường cao và di chuyển càng nhiều km mỗi ngày thì mức độ thôi thúc chuyển đổi sang phương tiện không phát thải càng lớn.

#### ⚙️ Đặc trưng 5: Nghịch đảo Nỗi lo Cự ly Pin (`Range_Anxiety_Inv`)
$$\text{Range\_Anxiety\_Inv} = \frac\frac{1}{\text{Range\_Anxiety\_Level\_Numeric} + 1.0}$$
* **Ý nghĩa:** Chuyển đổi biến lo ngại pin (Low=0, Medium=1, High=2) thành chỉ số nghịch đảo: giá trị càng cao thể hiện sự tự tin và an tâm về dung lượng pin của xe điện.

### 3. Từ điển Kỹ thuật Tiền xử lý trong `ColumnTransformer`

```
┌────────────────────────────────────────────────────────────────────────────────────────┐
│ KIẾN TRÚC TIỀN XỬ LÝ CHUẨN MỰC COLUMNTANSFORMER (21 INPUTS -> 25 OUTPUTS)              │
└────────────────────────────────────────────────────────────────────────────────────────┘
  │
  ├── 1. OneHotEncoder(drop='first'): Dành cho biến danh định (Nominal)
  │      Gender, City_Type, Current_Car_Type (3 biến -> 5 cột nhị phân)
  │
  ├── 2. OneHotEncoder(drop='if_binary'): Dành cho biến nhị phân Yes/No
  │      Home_Charging_Possible, Subsidy_Available (2 biến -> 2 cột nhị phân 0/1)
  │
  ├── 3. OrdinalEncoder: Dành cho biến thứ bậc
  │      Range_Anxiety_Level: ['Low', 'Medium', 'High'] -> [0, 1, 2] (1 cột)
  │
  ├── 4. RobustScaler(): Dành cho biến số lệch phải & nhiều giá trị ngoại lai
  │      Annual_Income_USD, Income_Per_Car, Commute_Burden, Subsidy_Income_Interaction (4 cột)
  │
  └── 5. StandardScaler(): Dành cho biến số có phân phối chuẩn / đối xứng
         Age, Daily_Commute_km, Environmental_Concern_Level, Charging Stations... (11 cột)
```

#### 📌 `RobustScaler`
* **Nó là gì?**  
  Phép chuẩn hóa dữ liệu dựa trên **Trung vị (Median)** và **Khoảng tứ phân vị (Interquartile Range - IQR)**:
  $$x_{\text{scaled}} = \frac\frac{x - Q_2}{Q_3 - Q_1} = \frac\frac{x - \text{Median}}{\text{IQR}}$$
* **Tại sao dùng cho Thu nhập và Biến tương tác thu nhập?**  
  Trung vị và IQR không bị ảnh hưởng bởi các giá trị ngoại lai cực lớn (Outliers). Nếu dùng StandardScaler, một vài triệu phú với thu nhập $500k sẽ kéo lệch giá trị Mean lên cao, làm nén chặt 99% người dùng bình thường lại một góc hẹp. RobustScaler giữ nguyên cấu trúc phân phối của nhóm đại đa số.

#### 📌 `StandardScaler`
* **Nó là gì?**  
  Phép chuẩn hóa đưa dữ liệu về phân phối chuẩn chuẩn hóa có trung bình $\mu = 0$ và độ lệch chuẩn $\sigma = 1$:
  $$z = \frac\frac{x - \mu}{\sigma}$$
* **Tại sao dùng?**  
  Thích hợp nhất cho các biến có phân phối dạng đối xứng (Tuổi, Điểm môi trường, Số lượng xe sở hữu), giúp các thuật toán tối ưu hóa bằng độ dốc (Gradient Descent) hội tụ nhanh hơn.

#### 📌 `OneHotEncoder(drop='first')` và Bẫy Biến Giả (Dummy Variable Trap)
* **Nó là gì?**  
  Nếu một biến có $k$ danh mục (ví dụ `City_Type` có 3 giá trị: Urban, Suburban, Rural), nếu tạo đủ 3 cột thì tổng 3 cột luôn bằng 1 ($\text{Urban} + \text{Suburban} + \text{Rural} = 1$). Điều này gây ra hiện tượng **Đa cộng tuyến hoàn hảo** với cột hệ số chặn (Intercept).  
  $ightarrow$ Tùy chọn `drop='first'` sẽ bỏ đi cột đầu tiên làm mốc tham chiếu (Reference Category), chỉ giữ lại $k-1$ cột (2 cột cho City_Type).

Kết quả cuối cùng là bầu chọn đa số (Majority Voting) giữa các cây.

  - `n_estimators=100`: Số lượng cây con trong rừng (100 cây).
  - `max_depth=12`: Giới hạn độ sâu để kiểm soát bộ nhớ RAM.
  - `n_jobs=-1`: Tận dụng 100% nhân CPU đa luồng để huấn luyện song song 100 cây.

#### 🤖 Model 4: LightGBM Classifier (Histogram-based Gradient Boosting)
* **Bản chất thuật toán:**  
  Thuật toán Tăng cường Độ dốc phát triển bởi Microsoft, áp dụng 2 cải tiến đột phá:
  1. *GOSS (Gradient-based One-Side Sampling):* Giữ lại các mẫu có độ dốc gradient lớn (khó học) và lấy mẫu ngẫu nhiên các mẫu có gradient nhỏ.
  2. *EFB (Exclusive Feature Bundling):* Gom các đặc trưng thưa ít khi cùng xuất hiện thành một đặc trưng duy nhất.
  3. *Cơ chế phân nhánh theo lá (`leaf-wise`):* Thay vì chia đều theo từng tầng (`depth-wise`), LightGBM luôn chọn lá có độ giảm hàm tổn thất lớn nhất để chia tiếp, giúp giảm sai số nhanh hơn rất nhiều.
* **Siêu tham số cấu hình:**
  - `n_estimators=1000`: 1000 cây tăng cường.
  - `learning_rate=0.05`: Tốc độ học nhỏ giúp mô hình hội tụ từ từ và bền vững.
  - `num_leaves=31`: Số lượng nút lá tối đa trên mỗi cây con ($2^5 - 1$).
  - `scale_pos_weight=4.73`: Trọng số tăng cường độ dốc cho nhãn dương $pprox \frac\frac{N_{\text{negative}}}{N_{\text{positive}}}$.
  - `verbose=-1`: Tắt thông báo hệ thống để log chạy sạch sẽ.

#### 🤖 Model 5: XGBoost Classifier (Extreme Gradient Boosting)
* **Bản chất thuật toán:**  
  Thuật toán Tăng cường Độ dốc tối ưu hóa bởi Tianqi Chen. Khác với Gradient Boosting truyền thống chỉ dùng đạo hàm bậc 1, XGBoost sử dụng **khai triển Taylor bậc hai** của hàm mất mát:
  $$\mathcal{L}^{(t)} pprox \sum_{i=1}^n \left[ l(y_i, \hat{y}_i^{(t-1)}) + g_i f_t(x_i) + \frac\frac{1}{2} h_i f_t^2(x_i) ight] + \Omega(f_t)$$
  Trong đó $g_i$ là đạo hàm bậc 1 (Gradient), $h_i$ là đạo hàm bậc 2 (Hessian), và $\Omega(f_t) = \gamma T + \frac\frac{1}{2} \lambda \sum w_j^2$ là thành phần phạt chính quy hóa $L_1$ và $L_2$ chống quá khớp.
* **Siêu tham số cấu hình:**
  - `tree_method='hist'`: Xây dựng cây dựa trên Histogram (rời rạc hóa biến liên tục thành các thùng bin), giúp tăng tốc độ tính toán gấp 10 lần trên dữ liệu nửa triệu dòng.
  - `max_depth=6`: Độ sâu tối đa của mỗi cây.
  - `scale_pos_weight=4.73`: Cân bằng trọng số phạt mất cân bằng nhãn.

### 2. Từ điển 6 Thước đo Đánh giá Kỹ thuật trong Bảng Leaderboard

#### 📌 1. Độ chính xác tổng quát (Accuracy)
$$\text{Accuracy} = \frac\frac{\text{TP} + \text{TN}}{\text{TP} + \text{TN} + \text{FP} + \text{FN}}$$
* **Nhược điểm:** Bị bóp méo nghiêm trọng khi nhãn mất cân bằng. Không được dùng làm tiêu chí chọn mô hình.

#### 📌 2. Độ chuẩn xác (Precision - Positive Class)
$$\text{Precision} = \frac\frac{\text{TP}}{\text{TP} + \text{FP}}$$
* **Ý nghĩa nghiệp vụ:** Trong 100 khách hàng mà mô hình dự đoán là sẽ mua xe điện, có bao nhiêu người thực sự mua? Precision cao giúp tránh lãng phí chi phí marketing cho người không có nhu cầu.

#### 📌 3. Độ nhạy / Tỷ lệ phát hiện (Recall / Sensitivity)
$$\text{Recall} = \frac\frac{\text{TP}}{\text{TP} + \text{FN}}$$
* **Ý nghĩa nghiệp vụ:** Trong 100 khách hàng thực tế có ý định mua xe ngoài thị trường, mô hình bắt trọn được bao nhiêu người? Recall cao giúp đại lý không bỏ lỡ khách hàng tiềm năng vào tay đối thủ. Cả 5 mô hình đều đạt Recall rất cao ($> 90\%$) nhờ tham số `scale_pos_weight`.

#### 📌 4. Điểm F1 (F1-Score)
$$F_1 = 2 \times \frac\frac{\text{Precision} \times \text{Recall}}{\text{Precision} + \text{Recall}}$$
* **Ý nghĩa:** Trung bình điều hòa giữa Precision và Recall. F1 chỉ đạt giá trị cao khi cả Precision và Recall đều cân bằng.

#### 📌 5. Diện tích dưới Đường cong ROC (ROC-AUC)
* **Ý nghĩa:** Đồ thị biểu diễn mối quan hệ giữa **True Positive Rate (Recall)** và **False Positive Rate ($\text{FPR} = \frac\frac{\text{FP}}{\text{FP}+\text{TN}}$)** trên toàn bộ dải ngưỡng từ 0 đến 1. Đo lường năng lực xếp hạng tổng thể. Đường ngẫu nhiên có $\text{AUC} = 0,5000$.

#### 📌 6. Diện tích dưới Đường cong Precision-Recall (PR-AUC / Average Precision)
* **Tại sao là thước đo quan trọng nhất?**  
  Khác với ROC-AUC sử dụng mẫu số có chứa $\text{TN}$ (vốn rất lớn trong dữ liệu mất cân bằng), **PR-AUC chỉ tập trung duy nhất vào lớp dương (khách hàng mua xe `Yes`)**. PR-AUC phản ánh chân thực nhất sự phân hóa năng lực giữa các thuật toán và là tiêu chí số 1 để xếp hạng Leaderboard!

| **Accuracy** | Overall Correct Ratio | $(\text{TP}+\text{TN})/\text{Total}$. Không dùng làm tiêu chí xếp hạng vì cạm bẫy mất cân bằng nhãn. |

| **PR Curve** | Precision-Recall Curve | Đường cong biểu diễn Precision theo Recall. Thước đo trọng yếu nhất cho dữ liệu mất cân bằng. |
